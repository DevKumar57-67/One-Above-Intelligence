*0.2.7 — CPU vs GPU

This is something you'll need to understand before Deep Learning.*

```python
'''
CPU

Designed for general-purpose sequential/parallel computation.

CPU
├── Core
├── Core
├── Core
└── Core

'''
```

```python
'''
GPU

Designed for massive parallel computation.

GPU
├── thousands of smaller processing units
├── massive parallelism
└── high-throughput matrix operations

'''
```

```python
'''
And neural networks are heavily based on:

Matrix multiplication
+
Tensor operations
'''

'''
Therefore GPUs become extremely useful for:

Deep Learning
CNNs
Transformers
LLMs
Diffusion
Fine-tuning

We'll go deep into this later.
'''
```
